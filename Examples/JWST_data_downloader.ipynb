# JWST Data Download and Reprocessing

This notebook downloads public JWST IFU spectroscopy from MAST and reruns the calibration pipeline with the settings used in the analysis:

1. **Detector1Pipeline** — UNCAL → RATE (standard ramp fitting).
2. **Spec2Pipeline** — RATE → CAL, with pixel-based background subtraction skipped so the diffuse sky signal is preserved.
3. **Spec3Pipeline** — combined S3D cube and X1D spectrum, with the master-background step skipped and local background subtraction enabled for point-like sources.

All of the logic lives in `data_downloader.py`; this notebook only configures and runs it.

BEFORE RUNNING:
Make sure you have the both jwst and astroquery installed! They can be installed via pip, "pip install jwst; pip install astroquery".

```python
from data_downloader import configure_crds, process_observation
```

## CRDS configuration

The JWST pipeline needs a CRDS reference-file cache. Point this at a **writable directory with plenty of space** — several tens of GB of reference files may be downloaded on first use.

```python
crds_cache = "/path/to/crds_cache"

configure_crds(crds_cache)
```

## Download and process one observation

Set the MAST observation ID, a base download directory, and the instrument (`"NIRSPEC/IFU"` or `"MIRI/IFU"`). The final level-3 products are written to `directory/mastDownload/JWST/<obs_id>/`.

If the notebook is interrupted, simply rerun this cell: completed Detector1 and Spec2 stages are detected and skipped.

```python
candidate_obs_id = "jw03131-o001_t001_nirspec_g140h-f100lp"
directory = "/path/to/download_directory"
instrument = "NIRSPEC/IFU"

folder = process_observation(candidate_obs_id, directory, instrument)

print("Level-3 products written to:", folder)
```
