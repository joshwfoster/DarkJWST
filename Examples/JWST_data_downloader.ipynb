# JWST Data Download and Reprocessing

This notebook downloads public JWST IFU spectroscopy from MAST and reruns the calibration pipeline with the settings used in the analysis:

1. **Detector1Pipeline** — UNCAL → RATE (standard ramp fitting).
2. **Spec2Pipeline** — RATE → CAL, with pixel-based background subtraction skipped so the diffuse sky signal is preserved.
3. **Spec3Pipeline** — combined S3D cube and X1D spectrum, with the master-background step skipped and local background subtraction enabled for point-like sources.

All of the logic lives in `data_downloader.py`; this notebook only configures and runs it.

BEFORE RUNNING:
Make sure you have the both jwst and astroquery installed! They can be installed via pip, "pip install jwst; pip install astroquery".

```python
################################
###   Imports and Paths      ###
################################

import os
import sys
import warnings
from pathlib import Path

################################
###   Repository Paths       ###
################################

EXAMPLES_DIR = Path.cwd()
PROJECT_DIR = EXAMPLES_DIR.parent

PYTHON_DIR = PROJECT_DIR / "python"
DATA_DIR = PROJECT_DIR / "data"

IRFS_DIR = DATA_DIR / "IRFs"
DUST_DIR = DATA_DIR / "MWDUST"
EXAMPLE_DATA_DIR = DATA_DIR / "ExampleData"

os.environ["DUST_DIR"] = str(DUST_DIR)

if str(PYTHON_DIR) not in sys.path:
    sys.path.insert(0, str(PYTHON_DIR))

from data_downloader import configure_crds, change_nirspec_outer_bkg_radii, process_observation
```
